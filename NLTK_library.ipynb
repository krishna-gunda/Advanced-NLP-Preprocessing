#**Regex Tokenizer Example**

````python
# Import RegexpTokenizer from nltk
from nltk.tokenize import RegexpTokenizer

# Small paragraph example
text1 = """
Hello!!! My name is Krishna 😊.
I love AI, Machine Learning, and NLP.
Follow me @krishna #AI #MachineLearning
"""

# Why we use Regex Tokenizer:
# Regex tokenizer helps us extract clean words
# and remove punctuation automatically.

# Create tokenizer
# \w+ means take only words and numbers
tokenizer = RegexpTokenizer(r'\w+')

# Tokenize text
tokens = tokenizer.tokenize(text1)

# Print tokens
print(tokens)
````

#**Full POS-Based Lemmatization Pipeline**

````python
import nltk

from nltk.tokenize import word_tokenize
from nltk import pos_tag
from nltk.stem import WordNetLemmatizer

# Downloads
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('wordnet')

# Paragraph
text = "Krishna is running and studying NLP."

# Tokenize
words = word_tokenize(text)

# POS tagging
tags = pos_tag(words)

# Lemmatizer
lemmatizer = WordNetLemmatizer()

# Convert POS tags
def get_wordnet_pos(tag):

    if tag.startswith('V'):
        return 'v'

    elif tag.startswith('J'):
        return 'a'

    elif tag.startswith('R'):
        return 'r'

    else:
        return 'n'

# Lemmatization
lemmatized_words = []

for word, tag in tags:

    pos = get_wordnet_pos(tag)

    lemma = lemmatizer.lemmatize(word, pos)

    lemmatized_words.append(lemma)

print(lemmatized_words)
````

# **Difference Between Normal Lemmatization and POS-Based Lemmatization 📘**

| Feature | Normal Lemmatization | POS-Based Lemmatization |
|---|---|---|
| Uses POS Tags | ❌ No | ✅ Yes |
| Accuracy | Less Accurate | More Accurate |
| Understands Grammar | ❌ No | ✅ Yes |
| Default Assumption | Assumes word is noun | Uses given POS |
| Output Quality | Sometimes incorrect | More meaningful |
| Context Awareness | ❌ No | ✅ Yes |
| NLP Performance | Moderate | Better |
| Example | running → running | running → run |
| Example | better → better | better → good |
